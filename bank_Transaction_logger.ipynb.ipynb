```python
import time
# Part 1 - Scope Practice

# Global variable
total_balance = 0


def deposit(amount):
    global total_balance
    total_balance = total_balance + amount
    return total_balance


def withdraw(amount):
    global total_balance
    total_balance = total_balance - amount
    return total_balance


# Closure Counter
def make_counter():
    count = 0

    def inner():
        nonlocal count
        count = count + 1
        return count

    return inner


transaction_counter = make_counter()

# Part 2 - Decorators 

# Log Transaction Decorator
def log_transaction(function):

    def wrapper(*args, **kwargs):
        print("\n--- BEFORE TRANSACTION ---")
        print("Function:", function.__name__)
        print("Arguments:", args)

        result = function(*args, **kwargs)

        print("--- AFTER TRANSACTION ---")
        print("Result:", result)

        return result

    return wrapper


# Positive Amount Decorator
def require_positive_amount(function):

    def wrapper(*args, **kwargs):

        amount = args[0]

        if amount <= 0:
            print("\nERROR: Amount must be positive!")
            return None

        return function(*args, **kwargs)

    return wrapper


# Part 3 - Track Calls Decorator

def track_calls(function):

    count = 0

    def wrapper(*args, **kwargs):
        nonlocal count

        count = count + 1

        print("Call number:", count)

        return function(*args, **kwargs)

    return wrapper


# Part 4 - Timer Decorator

def timer(function):

    def wrapper(*args, **kwargs):

        start_time = time.time()

        result = function(*args, **kwargs)

        end_time = time.time()

        execution_time = end_time - start_time

        print("Execution time:", execution_time, "seconds")

        return result

    return wrapper 
# Apply Decorators

@track_calls
@timer
@log_transaction
@require_positive_amount
def deposit(amount):

    global total_balance

    total_balance = total_balance + amount

    transaction_counter()

    return total_balance


@track_calls
@timer
@log_transaction
@require_positive_amount
def withdraw(amount):

    global total_balance

    total_balance = total_balance - amount

    transaction_counter()

    return total_balance

# Main Program

print("===== BANK TRANSACTION LOGGER =====")

deposit(100000)

withdraw(100000)

deposit(500)

deposit(500)

withdraw(100)

withdraw(50)

print("\n===== FINAL DETAILS =====")

print("Final Total Balance:", total_balance)

print("Total Valid Transactions:", transaction_counter())

print("\nDeposit and Withdraw call counts are shown above by track_calls.")
```
